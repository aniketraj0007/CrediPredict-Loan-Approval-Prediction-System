### Step 2: Model Training Insights for Interviews

#### 1. Why Logistic Regression?
- Binary classification problem (Approved vs Rejected).
- Highly explainable linear decision boundary with calibrated probabilities.

#### 2. Why Decision Tree?
- Non-linear relationship modeling without strict distributional assumptions.
- Interpretable decision rules (if Credit_History == 1 and Total_Income > X).

#### 3. Why Random Forest?
- Ensemble method combining multiple decision trees to reduce variance and prevent overfitting.

#### 4. Why Stratified Train-Test Split?
- Preserves identical target proportions (`Y` vs `N`) across both training and test datasets.
