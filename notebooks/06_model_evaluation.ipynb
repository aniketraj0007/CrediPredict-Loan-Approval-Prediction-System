### Step 1: Load Model Comparison Metrics Report

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

report_path = '../reports/model_comparison.csv'
report_df = pd.read_csv(report_path)
report_df
```

--- 
### Step 2: Visualization of Performance Metrics across Models

```python
metrics_df = pd.melt(report_df, id_vars=['Model'], value_vars=['Accuracy', 'Precision', 'Recall', 'F1 Score', 'ROC-AUC'],
                     var_name='Metric', value_name='Score')

plt.figure(figsize=(10, 5))
sns.barplot(data=metrics_df, x='Metric', y='Score', hue='Model', palette='Set2')
plt.title('Classification Performance Comparison Across Models', fontsize=14, fontweight='bold')
plt.ylim(0, 1.1)
plt.ylabel('Score (0.0 - 1.0)', fontsize=12)
plt.legend(title='Machine Learning Model', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

### Step 3: Interview Explainability Guide for Classification Metrics

1. **Accuracy:** Measures total correct predictions (Approved + Rejected) divided by total applications.
2. **Precision:** Answers *"When the model predicts a loan is Approved, how often is it actually safe?"* Crucial for banks to minimize bad loan defaults.
3. **Recall:** Answers *"Out of all truly deserving applicants, what fraction did the model approve?"* Crucial to avoid rejecting creditworthy customers.
4. **F1 Score:** Harmonic mean of Precision and Recall. Balances default risk prevention with business growth.
5. **ROC-AUC:** Area under Receiver Operating Characteristic curve. Evaluates class discrimination capability across decision thresholds.
